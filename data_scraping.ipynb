```python
import os, shutil                     # os → papka/fayl bilan ishlash, shutil → fayllarni ko'chirish/nusxalash

source_dir = '/content/pilaf_images/uzbek pilaf plov food'   # Rasmlar hozir turgan papka (qayerdan)
target_dir = '/content/drive/MyDrive/food101_dataset/pilaf'  # Rasmlar ko'chiriladigan papka (qayerga) → Google Drive

os.makedirs(target_dir, exist_ok=True)   # Maqsad papkani yaratamiz; oldin bor bo'lsa xato bermaydi

for filename in os.listdir(source_dir):  # Manba papkadagi har bir fayl nomini birma-bir olamiz
    src_file = os.path.join(source_dir, filename)   # To'liq manzil: manba papka + fayl nomi
    dst_file = os.path.join(target_dir, filename)   # To'liq manzil: maqsad papka + fayl nomi
    shutil.move(src_file, dst_file)                 # Faylni manbadan maqsadga KO'CHIRAMIZ (manbada qolmaydi)
```
